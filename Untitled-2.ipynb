```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
labels=['young Adult','Adult','middle-aged','senio']
df['age_group']=pd.qcut(df['age'], q=4, labels= labels)
```

```python
frequency_mapping ={
    'fortnightly':14,
    'weekly' :7,
    'Monthly':30 ,
    'Quarterly':90,
    'Bi -weekly':14,
    'Annually':356,
    'Every 3 Months':90
}
df['purchase_frequency_day'] =df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine, text

engine = create_engine(
    "mysql+pymysql://root:deva%401122@localhost:3306/customer_behavior"
)

print("Engine created successfully!")
```

```python
try:
    with engine.connect() as connection:
        result = connection.execute(text("SELECT DATABASE()"))
        print("✅ Connected successfully!")
        print("Database:", result.scalar())

except Exception as e:
    print("❌ Connection failed:")
    print(e)
```

```python
with engine.connect() as connection:
    result = connection.execute(text("SHOW TABLES"))

    for row in result:
        print(row)
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("DESCRIBE india_job_market"))

    for row in result:
        print(row)
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("SELECT * FROM india_job_market LIMIT 10")
    )

    for row in result:
        print(row)
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("SELECT COUNT(*) FROM india_job_market")
    )

    print("Total rows:", result.scalar())
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("""
            SELECT *
            FROM india_job_market
            LIMIT 10
        """)
    )

    for row in result:
        print(row)
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("""
            SELECT AVG(Salary_LPA_INR) AS Average_Salary
            FROM india_job_market
        """)
    )

    print(result.fetchone())
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("""
            SELECT MAX(Salary_LPA_INR) AS Highest_Salary
            FROM india_job_market
        """)
    )

    print(result.fetchone())
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("""
            SELECT 
                City,
                ROUND(AVG(Salary_LPA_INR), 2) AS Average_Salary
            FROM india_job_market
            GROUP BY City
            ORDER BY Average_Salary DESC
        """)
    )

    for row in result:
        print(row)
```

```python
with engine.connect() as connection:
    result = connection.execute(
        text("""
            SELECT 
                Work_Mode,
                COUNT(*) AS Job_Count
            FROM india_job_market
            GROUP BY Work_Mode
            ORDER BY Job_Count DESC
        """)
    )

    for row in result:
        print(row)
```
